```python
import numpy as np
import math

# Without Kernel Trick :(((((((
```

```python
class SVR_Kinda:
    def __init__(self, X, y, lr=0.001, benchmark=0.00001, kernel_function="rbf", var: float=0.5, gamma: float=0.5):
        self.X = X
        self.y = y
        self.lr = lr
        self.benchmark = benchmark
        self.kernel_function = kernel_function;
        self.var = var
        self.gamma = gamma

        self.X = self.kernel_apply("rbf", self.X);
        
        self.w = np.random.randn(self.X.shape[1])

    def rbf_kernel(self, X: np.ndarray) -> np.ndarray:
        X_norm = np.sum(X ** 2, axis = -1)
        K = self.var * np.exp(-self.gamma * (X_norm[:,None] + X_norm[None,:] - 2 * np.dot(X, X.T)))
        return K

    def kernel_apply(self, kernel_func, X):
        if(kernel_func == "rbf"):
            X = self.rbf_kernel(self.X)
        else:
            X = kernel_func(self.X)
        return X;
                
    def predict(self, X):
        X = self.kernel_apply("rbf", X)
        return np.dot(X, self.w)
    
    def error(self, X, y):
        return abs((self.predict(X) - y));
    
    def calc_grad(self, X, y):
        grads = np.dot(X.T, self.error(self.X, self.y))
        return grads
    
    def update_weights(self, grads):
        for i in range(len(self.w)):
            self.w[i] -= self.lr * grads[i]
    
    def train(self):
        min_error = self.error(self.X, self.y).mean();
        
        cnt_err_bigger_than_pred = 0;
        
        while(True):
            self.update_weights(self.calc_grad(self.X, self.y))
            error = self.error(self.X, self.y).mean()
            
            if(min_error - error < self.benchmark): 
                cnt_err_bigger_than_pred += 1;
            else:
                min_error = error;
                cnt_err_bigger_than_pred = 0;
            
            if(cnt_err_bigger_than_pred > 10): break
            
        print("Min Error: ", min_error);
        print("Weights: ", self.w)
        
```
