```python

from functools import lru_cache

lists = ((12, 6, 3, 6), (13, 8, 11, 7), (4, 1, 8, 7), (12, 14, 11, 4))
l = [np.flip(x) for x in lists]

lists=l

@lru_cache(None)
def game(i, j, k, l):
    if(i == 4 and j == 4 and k == 4 and l == 4):
        return 0
        
    move_i = -1e18
    move_j = -1e18
    move_k = -1e18
    move_l = -1e18

    if(i < 4): move_i = lists[0][i] - (game(i + 1, j, k, l))
    if(j < 4): move_j = lists[1][j] - (game(i, j + 1, k, l))
    if(k < 4): move_k = lists[2][k] - (game(i, j, k + 1, l))
    if(l < 4): move_l = lists[3][l] - (game(i, j, k, l + 1))

    # print(i, j, k, l, " --> ", max(move_i, move_j, move_k, move_l))

    return max(move_i, move_j, move_k, move_l)
```
