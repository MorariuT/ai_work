```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super(MyModel, self).__init__()

        self.lstm1 = nn.LSTM(768, hidden_size=1024, num_layers=2)
        self.linear1 = nn.Linear(in_features=1024, out_features=256)
        self.linear2 = nn.Linear(in_features=256, out_features=2)
        self.relu = nn.ReLU()

    def forward(self, X):
        self.out = self.lstm1(X)[0]
        self.out = self.relu(self.linear1(self.out))
        self.out = self.linear2(self.out)

        return self.out
```
