```python
import pandas as pd
```

```python
train = pd.read_csv("train_tables.csv")
```

```python
test = pd.read_csv("test_tables.csv")
```

```python
# train.dropna(inplace=True)
```

```python

def preproc(df: pd.DataFrame):
    drop_cols = ["feat_2", "hour", "minute"]
    df.drop(columns=drop_cols, inplace=True);
    for col in df.columns:
        df[col].fillna(df[col].mean(), inplace=True)


    return df;

train = preproc(train)
test = preproc(test)
```

```python
train.sort_values("target", inplace=True)
# test.sort_values("feat_1", inplace=True)

test_ids = test["id"]
test.drop(columns="id", inplace=True);
```

```python
train["feat_1"].ffill(inplace=True)
train["feat_1"].bfill(inplace=True)

# test.ffill(inplace=True)
# test.bfill(inplace=True)
```

```python
train.sort_index(inplace=True)
```
