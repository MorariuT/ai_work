```python
import pandas as pd
from tqdm import tqdm
tqdm.pandas()
```

```python
train = pd.read_json("train.json")
test = pd.read_json("test.json")
```

```python
train
```

```python
ingredients = {}
```

```python
def getIng(df: pd.DataFrame):
    def extract_ing(l: list):
        # global ingredients
        for elm in l: ingredients[elm] = 1
    df["ingredients"].apply(lambda x: extract_ing(x))
    
def putIng(df: pd.DataFrame):
    
    for i in ingredients:
        df[i] = 0;
    
    def putSkibidi(row):
        l = row["ingredients"]
        for elm in l: 
            row[elm] += 1;
        return row
    
    df = df.progress_apply(putSkibidi, axis=1)

    return df

getIng(train)
getIng(test)

train = putIng(train)
test = putIng(test)
```

```python
train.to_pickle("train_one_hot.pkl")
test.to_pickle("test_one_hot.pkl")
```
